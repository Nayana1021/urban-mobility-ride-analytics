# 2. Data Cleaning & Feature Engineering

The raw dataset contains 100,000 ride booking records for July 2024.

The cleaning process focuses on:

- Standardising column names and categorical values
- Converting date/time fields into appropriate formats
- Handling missing values based on business meaning
- Creating analytical flags for successful, cancelled and incomplete rides
- Validating numerical fields
- Creating derived metrics required for SQL and Power BI

The raw dataset is preserved without modification.

```python
# ============================================================
# DATA CLEANING & FEATURE ENGINEERING
# ============================================================

clean_df = df.copy()

# ------------------------------------------------------------
# 1. Standardise column names
# ------------------------------------------------------------

clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.replace(" ", "_")
)

# ------------------------------------------------------------
# 2. Standardise text columns
# ------------------------------------------------------------

text_columns = clean_df.select_dtypes(include="object").columns

for col in text_columns:
    clean_df[col] = clean_df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# ------------------------------------------------------------
# 3. Date and time
# ------------------------------------------------------------

clean_df["Date"] = pd.to_datetime(
    clean_df["Date"],
    errors="coerce"
)

clean_df["Time"] = pd.to_datetime(
    clean_df["Time"].astype(str),
    errors="coerce"
).dt.time

# Create useful date dimensions

clean_df["Day"] = clean_df["Date"].dt.day
clean_df["Day_Name"] = clean_df["Date"].dt.day_name()
clean_df["Week"] = clean_df["Date"].dt.isocalendar().week.astype(int)
clean_df["Month"] = clean_df["Date"].dt.month
clean_df["Hour"] = pd.to_datetime(
    clean_df["Time"].astype(str),
    errors="coerce"
).dt.hour

# ------------------------------------------------------------
# 4. Numeric columns
# ------------------------------------------------------------

numeric_columns = [
    "V_TAT",
    "C_TAT",
    "Booking_Value",
    "Ride_Distance",
    "Driver_Ratings",
    "Customer_Rating"
]

for col in numeric_columns:
    if col in clean_df.columns:
        clean_df[col] = pd.to_numeric(
            clean_df[col],
            errors="coerce"
        )

# ------------------------------------------------------------
# 5. Booking status flags
# ------------------------------------------------------------

clean_df["Is_Successful_Ride"] = (
    clean_df["Booking_Status"]
    .eq("Success")
    .astype(int)
)

clean_df["Is_Customer_Cancelled"] = (
    clean_df["Booking_Status"]
    .eq("Canceled by Customer")
    .astype(int)
)

clean_df["Is_Driver_Cancelled"] = (
    clean_df["Booking_Status"]
    .eq("Canceled by Driver")
    .astype(int)
)

clean_df["Is_Incomplete_Ride"] = (
    clean_df["Booking_Status"]
    .eq("Incomplete")
    .astype(int)
)

# ------------------------------------------------------------
# 6. Overall ride outcome
# ------------------------------------------------------------

clean_df["Ride_Outcome"] = np.select(
    [
        clean_df["Is_Successful_Ride"].eq(1),
        clean_df["Is_Customer_Cancelled"].eq(1),
        clean_df["Is_Driver_Cancelled"].eq(1),
        clean_df["Is_Incomplete_Ride"].eq(1)
    ],
    [
        "Completed",
        "Customer Cancelled",
        "Driver Cancelled",
        "Incomplete"
    ],
    default="Other"
)

# ------------------------------------------------------------
# 7. Revenue eligibility
# ------------------------------------------------------------

# Booking value is treated as realised ride value only for
# successfully completed rides.

clean_df["Realised_Revenue"] = np.where(
    clean_df["Is_Successful_Ride"].eq(1),
    clean_df["Booking_Value"],
    0
)

# ------------------------------------------------------------
# 8. Payment method quality flag
# ------------------------------------------------------------

clean_df["Payment_Method_Missing"] = (
    clean_df["Payment_Method"].isna().astype(int)
)

# ------------------------------------------------------------
# 9. Rating quality flags
# ------------------------------------------------------------

clean_df["Customer_Rating_Missing"] = (
    clean_df["Customer_Rating"].isna().astype(int)
)

clean_df["Driver_Rating_Missing"] = (
    clean_df["Driver_Ratings"].isna().astype(int)
)

# ------------------------------------------------------------
# 10. Ride distance quality
# ------------------------------------------------------------

clean_df["Ride_Distance_Missing"] = (
    clean_df["Ride_Distance"].isna().astype(int)
)

clean_df["Ride_Distance_Valid"] = (
    clean_df["Ride_Distance"].gt(0).astype(int)
)

# ------------------------------------------------------------
# 11. Final row count
# ------------------------------------------------------------

print("Cleaning and feature engineering completed.")
print(f"Rows: {len(clean_df):,}")
print(f"Columns: {len(clean_df.columns):,}")
```

```python
# ============================================================
# CLEANING VALIDATION
# ============================================================

validation = pd.DataFrame({
    "Metric": [
        "Raw Rows",
        "Clean Rows",
        "Raw Columns",
        "Clean Columns",
        "Duplicate Rows"
    ],
    "Value": [
        len(df),
        len(clean_df),
        len(df.columns),
        len(clean_df.columns),
        clean_df.duplicated().sum()
    ]
})

display(validation)

print("\nRide Outcome:")
display(
    clean_df["Ride_Outcome"]
    .value_counts(dropna=False)
    .to_frame("Bookings")
)

print("\nSuccessful Ride Rate:")
print(
    round(clean_df["Is_Successful_Ride"].mean() * 100, 2),
    "%"
)

print("\nRealised Revenue:")
print(
    f"₹{clean_df['Realised_Revenue'].sum():,.0f}"
)
```
